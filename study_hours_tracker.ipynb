```python
# ============================================
# STUDY HOURS TRACKER
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import json

# -----------------------------
# 1. Enter your study hours
# -----------------------------

days = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

hours = []

print("STUDY HOURS TRACKER")
print("-------------------")
print("Enter your study hours for each day.")

for day in days:
    while True:
        try:
            value = float(input(f"{day}: "))

            if value < 0:
                print("Please enter 0 or a positive number.")
            else:
                hours.append(value)
                break

        except ValueError:
            print("Please enter a valid number.")


# -----------------------------
# 2. Create the data
# -----------------------------

data = pd.DataFrame({
    "Day": days,
    "Study Hours": hours
})


# -----------------------------
# 3. Basic calculations
# -----------------------------

total_hours = data["Study Hours"].sum()
average_hours = data["Study Hours"].mean()

highest_index = data["Study Hours"].idxmax()
lowest_index = data["Study Hours"].idxmin()

highest_day = data.loc[highest_index, "Day"]
highest_hours = data.loc[highest_index, "Study Hours"]

lowest_day = data.loc[lowest_index, "Day"]
lowest_hours = data.loc[lowest_index, "Study Hours"]


# -----------------------------
# 4. Weekly goal
# -----------------------------

weekly_goal = float(input("\nEnter your weekly study goal (hours): "))

remaining_hours = max(weekly_goal - total_hours, 0)

goal_percentage = (total_hours / weekly_goal) * 100 if weekly_goal > 0 else 0

goal_percentage_display = min(goal_percentage, 100)


# -----------------------------
# 5. Performance report
# -----------------------------

print("\n================================")
print("       STUDY PERFORMANCE")
print("================================")

print(f"Total study hours: {total_hours:.2f}")
print(f"Average per day: {average_hours:.2f} hours")

print(f"Most study: {highest_day} - {highest_hours:.2f} hours")
print(f"Least study: {lowest_day} - {lowest_hours:.2f} hours")

print(f"Weekly goal: {weekly_goal:.2f} hours")
print(f"Goal progress: {goal_percentage:.2f}%")

if remaining_hours > 0:
    print(f"Remaining to reach goal: {remaining_hours:.2f} hours")
else:
    print("Weekly goal completed! 🎉")


# -----------------------------
# 6. Simple performance message
# -----------------------------

if goal_percentage >= 100:
    print("Performance: Excellent! You achieved your weekly goal.")
elif goal_percentage >= 75:
    print("Performance: Very good! You are close to your goal.")
elif goal_percentage >= 50:
    print("Performance: Good start. Try to stay consistent.")
else:
    print("Performance: Keep working on consistency.")


# -----------------------------
# 7. Show the complete table
# -----------------------------

print("\n================================")
print("         WEEKLY DATA")
print("================================")

display(data)


# -----------------------------
# 8. Bar chart
# -----------------------------

plt.figure(figsize=(10, 5))

plt.bar(data["Day"], data["Study Hours"])

plt.title("Study Hours by Day")
plt.xlabel("Day")
plt.ylabel("Study Hours")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -----------------------------
# 9. Line chart
# -----------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    data["Day"],
    data["Study Hours"],
    marker="o"
)

plt.title("Weekly Study Pattern")
plt.xlabel("Day")
plt.ylabel("Study Hours")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()


# -----------------------------
# 10. Save data as CSV
# -----------------------------

data.to_csv("study_hours.csv", index=False)

print("\nYour study data has been saved as: study_hours.csv")


# -----------------------------
# 11. Save data as JSON
# -----------------------------

study_data = {
    "weekly_goal": weekly_goal,
    "total_hours": total_hours,
    "average_hours": average_hours,
    "days": days,
    "hours": hours
}

with open("study_hours_data.json", "w") as file:
    json.dump(study_data, file, indent=4)

print("Your project data has also been saved as: study_hours_data.json")

print("\n================================")
print("       PROJECT COMPLETE")
print("================================")
```
